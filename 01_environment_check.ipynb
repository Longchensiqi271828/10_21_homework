```python
import sys
import torch

print("Python 版本：", sys.version)
print("Python 路径：", sys.executable)
print("PyTorch 版本：", torch.__version__)
print("NVIDIA CUDA 可用：", torch.cuda.is_available())
```

```python
x = torch.tensor([1.0, 2.0, 3.0])
y = x * 2 + 1
print(y)
```

```python
x = torch.tensor(3.0, requires_grad=True)
y = x ** 2
y.backward()
print(x.grad)
```
